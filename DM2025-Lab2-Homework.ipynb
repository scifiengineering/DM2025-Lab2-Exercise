I loaded the dataset from the provided **JSON and CSV files** (`final_posts.json`, `data_identification.csv`, `emotion.csv`). I performed text cleaning which involved:

*   Removing non-alphanumeric characters.

*   **Lemmatization**: Converting words to their base form using NLTK's WordNetLemmatizer.
*   Removing English stopwords.

### 1.2 Feature Engineering Steps

I transitioned from simple TF-IDF to **Sequence Tokenization** for my Deep Learning models.
*   **Tokenization**: I converted words into integer indices using a vocabulary size of 10,000.
*   **Padding**: I padded all sequences to a fixed length of 100 to ensure uniform input for the neural network.
*   **Label Encoding**: I one-hot encoded the target emotion labels.

### 1.3 Explanation of Your Model

I implemented a **Bidirectional LSTM (Long Short-Term Memory)** neural network as my primary custom architecture.
*   **Embedding Layer**: Maps word indices to dense vectors of dimension **50** (reduced from 100 to fight overfitting).
*   **Bidirectional LSTM**: Captures context from both past and future words in the sequence. I used **32 units**.
*   **GlobalMaxPool1D**: Reduces the sequence output to a single vector.
*   **Dense & Dropout**: A fully connected layer with ReLU activation, **L2 Regularization**, and **Dropout (0.6)** to prevent overfitting.
*   **Output Layer**: A Softmax layer to predict the probability of each of the 4 emotions.

3.  **Ensemble Learning**: I implemented a **Soft Voting Ensemble**. I combined the probability predictions of my fine-tuned DistilBERT and RoBERTa models. By averaging their outputs, I smoothed out individual model errors, leading to better generalization and my best leaderboard score.

# **Competition Code**

## 1. Preprocessing Steps

```python
import pandas as pd
import numpy as np
import re
import os
import json
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.model_selection import train_test_split

# Download necessary NLTK data
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    # Remove user handles
    text = re.sub(r'@\w+', '', text)
    # Remove URLs
    text = re.sub(r'http\S+', '', text)
    # Remove non-alphanumeric characters (except spaces)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # Lowercase
    text = text.lower()
    # Tokenize (simple split)
    words = text.split()
    # Remove stopwords and Lemmatize
    cleaned_words = [lemmatizer.lemmatize(word) for word in words if word not in stop_words]
    return ' '.join(cleaned_words)

# --- DATA LOADING LOGIC ---
DATA_DIR = 'data'
# External path for large JSON dataset
EXTERNAL_DATA_DIR = '/Users/scifithedev/Documents/Taiwan Masters/NTHU/2025 Fall/Data Mining - Concepts,Techniques, andApplications/Lab-2/dm-lab-2-private-competition'
FINAL_POSTS_PATH = os.path.join(EXTERNAL_DATA_DIR, 'final_posts.json')

# Fallback to local data dir if external not found
if not os.path.exists(FINAL_POSTS_PATH):
    FINAL_POSTS_PATH = os.path.join(DATA_DIR, 'final_posts.json')

ID_FILE_PATH = os.path.join(DATA_DIR, 'data_identification.csv')
EMOTION_FILE_PATH = os.path.join(DATA_DIR, 'emotion.csv')

def load_competition_data():
    print("Loading competition data...")
    
    # Check for existence
    if not os.path.exists(ID_FILE_PATH): raise FileNotFoundError(f"Missing {ID_FILE_PATH}")
    if not os.path.exists(EMOTION_FILE_PATH): raise FileNotFoundError(f"Missing {EMOTION_FILE_PATH}")
    if not os.path.exists(FINAL_POSTS_PATH): raise FileNotFoundError(f"Missing {FINAL_POSTS_PATH}")

    # 1. Load IDs and Splits
    id_df = pd.read_csv(ID_FILE_PATH)
    print(f"Loaded {len(id_df)} IDs from data_identification.csv")
    
    # 2. Load Labels (Training set only)
    emotion_df = pd.read_csv(EMOTION_FILE_PATH)
    print(f"Loaded {len(emotion_df)} labels from emotion.csv")
    
    # 3. Load Text (Parsing nested JSON)
    print(f"Loading text from {FINAL_POSTS_PATH}...")
    with open(FINAL_POSTS_PATH, 'r') as f:
        posts_data = json.load(f)
    
    # Parse the nested structure: root -> _source -> post -> [post_id, text]
    parsed_data = []
    for entry in posts_data:
        try:
            post_info = entry['root']['_source']['post']
            parsed_data.append({
                'id': post_info['post_id'],
                'text': post_info['text']
            })
        except KeyError:
            continue # Skip malformed entries
            
    text_df = pd.DataFrame(parsed_data)
    print(f"Loaded {len(text_df)} texts from final_posts.json")
    
    # --- MERGING ---
    # Ensure 'id' is string in all dfs
    id_df['id'] = id_df['id'].astype(str)
    emotion_df['id'] = emotion_df['id'].astype(str)
    text_df['id'] = text_df['id'].astype(str)

    # Merge text with IDs
    full_df = pd.merge(id_df, text_df, on='id', how='left')
    
    # Split into Train and Test based on 'split' column
    train_ids = full_df[full_df['split'] == 'train']
    test_ids = full_df[full_df['split'] == 'test']
    
    # Merge Labels into Train
    train_df = pd.merge(train_ids, emotion_df, on='id', how='left')
    
    # Test set for competition (no labels)
    competition_test_df = test_ids.copy()
    
    return train_df, competition_test_df

# Main Execution
try:
    train_df, competition_test_df = load_competition_data()
    
    # Split Train into Train/Validation
    train_df, val_df = train_test_split(train_df, test_size=0.2, random_state=42, stratify=train_df['emotion'])
    
    print(f"Data prepared successfully.")
    print(f"Training Set: {len(train_df)} samples")
    print(f"Validation Set: {len(val_df)} samples")
    print(f"Competition Test Set: {len(competition_test_df)} samples (blind)")

    # Apply cleaning
    print("Cleaning training data...")
    train_df['cleaned_text'] = train_df['text'].apply(clean_text)
    print("Cleaning validation data...")
    val_df['cleaned_text'] = val_df['text'].apply(clean_text)
    print("Cleaning competition test data...")
    competition_test_df['cleaned_text'] = competition_test_df['text'].apply(clean_text)
    
    print("Preprocessing complete.")
    print(train_df[['text', 'cleaned_text']].head())

except Exception as e:
    print(f"Error: {e}")
    print("Please ensure all data files are present.")
```

## 2. Feature Engineering Steps

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical
from transformers import DistilBertTokenizer
import numpy as np

# Parameters
MAX_WORDS = 10000
MAX_LEN_LSTM = 100
MAX_LEN_BERT = 128
MODEL_NAME = 'distilbert-base-uncased'

# Alias val_df to test_df for compatibility
if 'val_df' in locals():
    test_df = val_df
    print("Using 'val_df' as 'test_df' for validation.")

# --- PREPARATION FOR LSTM ---
print("\n--- Preparing Data for LSTM ---")
# Tokenization
tokenizer_lstm = Tokenizer(num_words=MAX_WORDS, oov_token='<OOV>')
tokenizer_lstm.fit_on_texts(train_df['cleaned_text'])

# Convert text to sequences
X_train_seq = tokenizer_lstm.texts_to_sequences(train_df['cleaned_text'])
X_test_seq = tokenizer_lstm.texts_to_sequences(test_df['cleaned_text'])

# Padding
X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LEN_LSTM, padding='post', truncating='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=MAX_LEN_LSTM, padding='post', truncating='post')

print("LSTM Data Ready.")
print("X_train_pad shape:", X_train_pad.shape)

# --- PREPARATION FOR BERT ---
print("\n--- Preparing Data for BERT ---")
print(f"Loading {MODEL_NAME} tokenizer...")
tokenizer_bert = DistilBertTokenizer.from_pretrained(MODEL_NAME)

def bert_encode(data, maximum_length):
    input_ids = []
    attention_masks = []

    for text in data:
        encoded = tokenizer_bert.encode_plus(
            text,
            add_special_tokens=True,
            max_length=maximum_length,
            padding='max_length',
            return_attention_mask=True,
            truncation=True
        )
        input_ids.append(encoded['input_ids'])
        attention_masks.append(encoded['attention_mask'])
        
    return np.array(input_ids), np.array(attention_masks)

print("Tokenizing data for BERT (using raw text)...")
# Use raw 'text' for BERT, not 'cleaned_text'
X_train_ids, X_train_masks = bert_encode(train_df['text'], MAX_LEN_BERT)
X_test_ids, X_test_masks = bert_encode(test_df['text'], MAX_LEN_BERT)

print("BERT Data Ready.")
print("X_train_ids shape:", X_train_ids.shape)

# --- LABELS ---
# Encode labels (Shared)
label_encoder = LabelEncoder()
y_train_le = label_encoder.fit_transform(train_df['emotion'])
y_test_le = label_encoder.transform(test_df['emotion'])

# Convert to categorical
y_train_cat = to_categorical(y_train_le)
y_test_cat = to_categorical(y_test_le)

print("\nLabels encoded.")
print("Classes:", label_encoder.classes_)
```

## 3. Model Implementation Steps

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Bidirectional, Dropout, GlobalMaxPool1D
from tensorflow.keras.callbacks import EarlyStopping as EarlyStopping_Keras
from tensorflow.keras import regularizers
from transformers import TFDistilBertForSequenceClassification
from sklearn.utils import class_weight
from sklearn.metrics import f1_score
import tensorflow as tf
import tf_keras
from tf_keras.callbacks import EarlyStopping as EarlyStopping_TF_Keras
from tf_keras.losses import CategoricalCrossentropy as CategoricalCrossentropy_TF_Keras
import numpy as np

# Optimizer Imports
# 1. For LSTM (Keras 3 / tensorflow.keras): Use standard Adam. Legacy is not supported in Keras 3.
from tensorflow.keras.optimizers import Adam as Adam_Keras
print("Using standard Adam optimizer for LSTM (Keras 3)")

# 2. For DistilBERT (Keras 2 / tf_keras): Use AdamW (Weight Decay) as recommended for Transformers.
# Try to import AdamW from tf_keras, falling back to Adam if not found.
try:
    from tf_keras.optimizers import AdamW as AdamW_TF_Keras
    print("Using tf_keras.optimizers.AdamW for DistilBERT")
    optimizer_bert_class = AdamW_TF_Keras
    optimizer_kwargs = {'weight_decay': 0.01}
except (ImportError, AttributeError):
    print("AdamW not found in tf_keras.optimizers. Falling back to Legacy Adam.")
    try:
        from tf_keras.optimizers.legacy import Adam as Adam_Legacy
        optimizer_bert_class = Adam_Legacy
        optimizer_kwargs = {}
    except (ImportError, AttributeError):
        from tf_keras.optimizers import Adam as Adam_TF_Keras
        optimizer_bert_class = Adam_TF_Keras
        optimizer_kwargs = {}

# Calculate class weights
class_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train_le),
    y=y_train_le
)
class_weights_dict = dict(enumerate(class_weights))
print("Class Weights:", class_weights_dict)
num_classes = len(label_encoder.classes_)

# --- MODEL 1: Bi-LSTM (Improved Regularization) ---
print("\n" + "="*30)
print("TRAINING MODEL 1: Bi-LSTM (Regularized)")
print("="*30)

# Reduced complexity to fight overfitting
model_lstm = Sequential([
    Embedding(input_dim=MAX_WORDS, output_dim=50), # Reduced from 100
    Bidirectional(LSTM(32, return_sequences=True)), # Reduced from 64
    GlobalMaxPool1D(),
    Dense(32, activation='relu', kernel_regularizer=regularizers.l2(0.01)), # Added L2 Reg
    Dropout(0.6), # High dropout
    Dense(num_classes, activation='softmax')
])

model_lstm.compile(optimizer=Adam_Keras(learning_rate=0.001),
                   loss='categorical_crossentropy',
                   metrics=['accuracy'])

# Set patience to 15 to ensure it trains for all epochs as requested, but keeps best weights
early_stopping_lstm = EarlyStopping_Keras(monitor='val_loss', patience=15, restore_best_weights=True, verbose=1)

history_lstm = model_lstm.fit(
    X_train_pad, y_train_cat,
    epochs=15,
    batch_size=32,
    validation_data=(X_test_pad, y_test_cat),
    class_weight=class_weights_dict,
    callbacks=[early_stopping_lstm],
    verbose=1
)

# Evaluate LSTM
y_pred_probs_lstm = model_lstm.predict(X_test_pad)
y_pred_classes_lstm = np.argmax(y_pred_probs_lstm, axis=1)
f1_lstm = f1_score(np.argmax(y_test_cat, axis=1), y_pred_classes_lstm, average='macro')
print(f"Bi-LSTM Test Mean F1 Score: {f1_lstm:.4f}")


# --- MODEL 2: DistilBERT (Using tf_keras for compatibility) ---
print("\n" + "="*30)
print("TRAINING MODEL 2: DistilBERT")
print("="*30)

try:
    model_bert = TFDistilBertForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=num_classes)
except (TypeError, ValueError) as e:
    print(f"Error loading with defaults: {e}. Retrying with use_safetensors=False...")
    model_bert = TFDistilBertForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=num_classes, use_safetensors=False)

# Use the selected optimizer class for BERT
optimizer_bert = optimizer_bert_class(learning_rate=2e-5, epsilon=1e-08, **optimizer_kwargs)
loss_bert = CategoricalCrossentropy_TF_Keras(from_logits=True)

model_bert.compile(optimizer=optimizer_bert, loss=loss_bert, metrics=['accuracy'])

early_stopping_bert = EarlyStopping_TF_Keras(monitor='val_loss', patience=2, restore_best_weights=True, verbose=1)

history_bert = model_bert.fit(
    [X_train_ids, X_train_masks],
    y_train_cat,
    epochs=3,
    batch_size=16,
    validation_data=([X_test_ids, X_test_masks], y_test_cat),
    class_weight=class_weights_dict,
    callbacks=[early_stopping_bert],
    verbose=1
)

# Evaluate BERT
y_pred_logits_bert = model_bert.predict([X_test_ids, X_test_masks]).logits
y_pred_classes_bert = np.argmax(y_pred_logits_bert, axis=1)
f1_bert = f1_score(np.argmax(y_test_cat, axis=1), y_pred_classes_bert, average='macro')
print(f"DistilBERT Test Mean F1 Score: {f1_bert:.4f}")

# Comparison
print("\n" + "="*30)
print("MODEL COMPARISON")
print("="*30)
print(f"Bi-LSTM F1: {f1_lstm:.4f}")
print(f"DistilBERT F1: {f1_bert:.4f}")

if f1_bert > f1_lstm:
    print("Best Model: DistilBERT")
    best_model = model_bert
    best_model_type = 'bert'
else:
    print("Best Model: Bi-LSTM")
    best_model = model_lstm
    best_model_type = 'lstm'
```

```python
import matplotlib.pyplot as plt

def plot_history(history, model_name):
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Train Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title(f'{model_name} Accuracy')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Train Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title(f'{model_name} Loss')
    plt.ylabel('Loss')
    plt.xlabel('Epoch')
    plt.legend()

    plt.show()

# Plot Bi-LSTM
if 'history_lstm' in locals():
    print("Plotting Bi-LSTM History:")
    plot_history(history_lstm, "Bi-LSTM")
else:
    print("Bi-LSTM history not found.")

# Plot DistilBERT
if 'history_bert' in locals():
    print("Plotting DistilBERT History:")
    plot_history(history_bert, "DistilBERT")
else:
    print("DistilBERT history not found.")
```

```python
# 4.2 RoBERTa Model (Advanced Experiment)
# ---------------------------------------------------------
# RoBERTa (Robustly optimized BERT approach) often outperforms DistilBERT.
# We use the same training pipeline but with the RoBERTa tokenizer and model.

from transformers import TFRobertaForSequenceClassification, RobertaTokenizer
import tensorflow as tf
import numpy as np
from sklearn.metrics import f1_score

# Check for tf_keras (required for Transformers with TF 2.16+)
try:
    import tf_keras
    HAS_TF_KERAS = True
    print("tf_keras library is available. Using it for Model, Optimizer, and Callbacks.")
except ImportError:
    HAS_TF_KERAS = False
    print("tf_keras library NOT found. Using standard tensorflow.keras.")

MODEL_NAME_ROBERTA = 'roberta-base'

# --- PREPARATION FOR ROBERTA ---
print(f"Loading {MODEL_NAME_ROBERTA} tokenizer...")
tokenizer_roberta = RobertaTokenizer.from_pretrained(MODEL_NAME_ROBERTA)

def roberta_encode(data, maximum_length):
    input_ids = []
    attention_masks = []

    for text in data:
        encoded = tokenizer_roberta.encode_plus(
            text,
            add_special_tokens=True,
            max_length=maximum_length,
            padding='max_length',
            return_attention_mask=True,
            truncation=True
        )
        input_ids.append(encoded['input_ids'])
        attention_masks.append(encoded['attention_mask'])
        
    return np.array(input_ids), np.array(attention_masks)

print("Tokenizing data for RoBERTa...")
# Re-using MAX_LEN_BERT from previous cells (128)
X_train_ids_rob, X_train_masks_rob = roberta_encode(train_df['text'], MAX_LEN_BERT)
X_test_ids_rob, X_test_masks_rob = roberta_encode(test_df['text'], MAX_LEN_BERT)

print("RoBERTa Data Ready.")

# Initialize RoBERTa model
try:
    model_roberta = TFRobertaForSequenceClassification.from_pretrained(MODEL_NAME_ROBERTA, num_labels=num_classes)
except (TypeError, ValueError) as e:
    print(f"Error loading with defaults: {e}. Retrying with use_safetensors=False...")
    model_roberta = TFRobertaForSequenceClassification.from_pretrained(MODEL_NAME_ROBERTA, num_labels=num_classes, use_safetensors=False)

# Optimizer & Callbacks Configuration
# LOWER LEARNING RATE for RoBERTa (1e-5) to prevent overfitting
if HAS_TF_KERAS:
    # Use tf_keras components to match the model
    try:
        from tf_keras.optimizers import AdamW
        print("Using tf_keras.optimizers.AdamW (Weight Decay)")
        optimizer_roberta = AdamW(learning_rate=1e-5, weight_decay=0.01, epsilon=1e-08)
    except (ImportError, AttributeError):
        try:
            from tf_keras.optimizers.legacy import Adam
            print("Using tf_keras.optimizers.legacy.Adam (M1/M2 Acceleration)")
        except (ImportError, AttributeError):
            from tf_keras.optimizers import Adam
            print("Using tf_keras.optimizers.Adam")
        optimizer_roberta = Adam(learning_rate=1e-5, epsilon=1e-08)
    
    loss_roberta = tf_keras.losses.CategoricalCrossentropy(from_logits=True)
    
    # IMPORTANT: Must use tf_keras callback with tf_keras model
    early_stopping_roberta = tf_keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=2, restore_best_weights=True
    )
else:
    # Fallback to standard Keras
    try:
        from tensorflow.keras.optimizers import AdamW
        print("Using tensorflow.keras.optimizers.AdamW")
        optimizer_roberta = AdamW(learning_rate=1e-5, weight_decay=0.01, epsilon=1e-08)
    except (ImportError, AttributeError):
        try:
            from tensorflow.keras.optimizers.legacy import Adam
            print("Using tensorflow.keras.optimizers.legacy.Adam")
        except (ImportError, AttributeError):
            from tensorflow.keras.optimizers import Adam
            print("Using tensorflow.keras.optimizers.Adam")
        optimizer_roberta = Adam(learning_rate=1e-5, epsilon=1e-08)
        
    loss_roberta = tf.keras.losses.CategoricalCrossentropy(from_logits=True)
    
    early_stopping_roberta = tf.keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=2, restore_best_weights=True
    )

model_roberta.compile(optimizer=optimizer_roberta, loss=loss_roberta, metrics=['accuracy'])

print("\nTraining RoBERTa Model...")
history_roberta = model_roberta.fit(
    [X_train_ids_rob, X_train_masks_rob],
    y_train_cat,
    validation_data=([X_test_ids_rob, X_test_masks_rob], y_test_cat),
    epochs=4, # Increased epochs slightly since LR is lower
    batch_size=16,
    callbacks=[early_stopping_roberta],
    class_weight=class_weights_dict
)

# Evaluation
print("\nEvaluating RoBERTa Model...")
loss_rob, acc_rob = model_roberta.evaluate([X_test_ids_rob, X_test_masks_rob], y_test_cat)
print(f"RoBERTa Accuracy: {acc_rob:.4f}")

# F1 Score Calculation
y_pred_logits_rob = model_roberta.predict([X_test_ids_rob, X_test_masks_rob]).logits
y_pred_classes_rob = np.argmax(y_pred_logits_rob, axis=1)

f1_rob = f1_score(y_test_le, y_pred_classes_rob, average='macro')
print(f"RoBERTa Macro F1 Score: {f1_rob:.4f}")

# --- ENSEMBLE EVALUATION ---
print("\n" + "="*30)
print("ENSEMBLE EVALUATION (DistilBERT + RoBERTa)")
print("="*30)

# Get probabilities (Softmax)
probs_bert = tf.nn.softmax(y_pred_logits_bert).numpy()
probs_rob = tf.nn.softmax(y_pred_logits_rob).numpy()

# Average Probabilities
probs_ensemble = (probs_bert + probs_rob) / 2
pred_ensemble = np.argmax(probs_ensemble, axis=1)

f1_ensemble = f1_score(y_test_le, pred_ensemble, average='macro')
print(f"Ensemble Macro F1 Score: {f1_ensemble:.4f}")

# Compare and Select Best
scores = {'bert': f1_bert, 'roberta': f1_rob, 'ensemble': f1_ensemble}
best_type = max(scores, key=scores.get)

print(f"\nWINNER: {best_type.upper()} with F1: {scores[best_type]:.4f}")

best_model_type = best_type
if best_type == 'bert':
    best_model = model_bert
elif best_type == 'roberta':
    best_model = model_roberta
else:
    # For ensemble, we need both. We'll handle this in the submission cell.
    best_model = None 
    print("Ensemble selected. Both models will be used for submission.")
```
